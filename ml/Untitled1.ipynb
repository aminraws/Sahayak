```python
import pandas as pd
import numpy as np

np.random.seed(42)
```

```python
n = 1000
```

```python
worker_id = np.arange(1, n + 1)
```

```python
skills = np.random.choice(
    [
        "Plumbing",
        "Electrical",
        "Carpentry",
        "Cleaning",
        "Painting",
        "Gardening"
    ],
    size=n
)
```

```python
experience = np.random.randint(1, 21, size=n)
```

```python
rating = np.round(
    np.random.uniform(3.0, 5.0, size=n),
    1
)
```

```python
availability = np.random.choice(
    [0, 1],
    size=n,
    p=[0.25, 0.75]
)
```

```python
workload = np.random.randint(0, 11, size=n)
```

```python
distance = np.round(
    np.random.uniform(0.5, 20.0, size=n),
    2
)
```

```python
skill_match = np.random.choice(
    [0, 1],
    size=n,
    p=[0.2, 0.8]
)
```

```python
df = pd.DataFrame({
    "worker_id": worker_id,
    "skill": skills,
    "skill_match": skill_match,
    "distance": distance,
    "availability": availability,
    "experience": experience,
    "rating": rating,
    "workload": workload
})
```

```python
df.head()
```

```python
df["experience_score"] = df["experience"] / 20
```

```python
df["rating_score"] = df["rating"] / 5
```

```python
df["distance_score"] = 1 - (df["distance"] / 20)
```

```python
df["workload_score"] = 1 - (df["workload"] / 10)
```

```python
df["match_score"] = (
    df["skill_match"] * 0.35 +
    df["distance_score"] * 0.20 +
    df["availability"] * 0.15 +
    df["experience_score"] * 0.10 +
    df["rating_score"] * 0.10 +
    df["workload_score"] * 0.10
)
```

```python
df["match_score"] = df["match_score"] * 100
```

```python
np.random.seed(42)

probability = (
    0.35 * df["skill_match"] +
    0.20 * df["distance_score"] +
    0.15 * df["availability"] +
    0.10 * df["experience_score"] +
    0.10 * df["rating_score"] +
    0.10 * df["workload_score"]
)

df["selected"] = np.random.binomial(
    1,
    probability
)
```
