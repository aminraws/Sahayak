```python
import pandas as pd
import numpy as np
```

```python
n = 1000
```

```python
worker_id = [f"W{i:04d}" for i in range(1, n + 1)]
```

```python
first_names = [
    "Rajesh", "Amit", "Suresh", "Ravi", "Manoj",
    "Vikas", "Arun", "Deepak", "Pankaj", "Anil",
    "Sunil", "Rohit", "Karan", "Mohit", "Vivek"
]

last_names = [
    "Kumar", "Sharma", "Singh", "Verma", "Gupta",
    "Yadav", "Agarwal", "Patel", "Mishra", "Joshi"
]

names = [
    f"{np.random.choice(first_names)} {np.random.choice(last_names)}"
    for _ in range(n)
]
```

```python
skills = np.random.choice(
    [
        "Plumbing",
        "Electrical",
        "Carpentry",
        "Cleaning",
        "Painting",
        "Gardening"
    ],
    size=n
)
```

```python
experience = np.random.randint(1, 21, size=n)
```

```python
rating = np.round(
    np.random.uniform(3.0, 5.0, size=n),
    1
)
```

```python
availability = np.random.choice(
    [0, 1],
    size=n,
    p=[0.2, 0.8]
)
```

```python
workload = np.random.randint(0, 11, size=n)
```

```python
latitude = np.round(
    np.random.uniform(28.40, 29.20, size=n),
    6
)
```

```python
longitude = np.round(
    np.random.uniform(76.80, 77.80, size=n),
    6
)
```

```python
df = pd.DataFrame({
    "worker_id": worker_id,
    "name": names,
    "skill": skills,
    "experience": experience,
    "rating": rating,
    "workload": workload,
    "availability": availability,
    "latitude": latitude,
    "longitude": longitude
})
```
